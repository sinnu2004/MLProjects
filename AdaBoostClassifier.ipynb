```python
import numpy as np
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import AdaBoostClassifier
```

```python
df = pd.read_csv("/content/data (1).csv")
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df = df.drop(columns='Unnamed: 32')
```

```python
df.head()
```

```python
df = df.drop(columns='id')
```

```python
df.head()
```

```python
weak_learn = DecisionTreeClassifier(max_depth=1)
```

```python
clsf = AdaBoostClassifier(estimator=weak_learn,n_estimators=2)
```

```python
x = df.drop(columns='diagnosis')
y = df['diagnosis']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
clsf.fit(x_train,y_train)
```

```python
pred = clsf.predict(x_test)
```

```python
pred
```

```python
print("Training Score",clsf.score(x_train,y_train)*100)
print("Testing Score",clsf.score(x_test,y_test)*100)
```

```python
print("alpha:",clsf.estimator_weights_)
```

```python
clsf.estimator_errors_
```
