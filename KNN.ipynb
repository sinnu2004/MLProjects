```python
df = df.drop(columns=['6th Stage'])
```

```python
df.head(1)
```

```python
x = df.drop(columns='Status')
y = df['Status']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
x_train
```

```python
x_train.info()
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
from sklearn.neighbors import KNeighborsClassifier
model = KNeighborsClassifier(n_neighbors=5)
model.fit(x_train,y_train)
```

```python
pred = model.predict(x_test)
```

```python
pred = pd.DataFrame(pred)
```

```python
pred
```

```python
print("Training Score :",model.score(x_train,y_train)*100)
print("Testing Score :",model.score(x_test,y_test)*100)
```

```python
from sklearn.metrics import confusion_matrix,classification_report
print("Confusion Matrix",confusion_matrix(y_test,pred))
```

```python
print("Classification Report",classification_report(y_test,pred))
```

```python
from sklearn.metrics import f1_score, recall_score, precision_score, accuracy_score

print("f1 Score :", f1_score(y_test, pred, pos_label='Dead'))
print("recall score:", recall_score(y_test, pred, pos_label='Dead'))
print("precision score:", precision_score(y_test, pred, pos_label='Dead'))
print("accuracy score:", accuracy_score(y_test, pred))
```
