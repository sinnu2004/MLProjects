```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
```

```python
from numpy._core.defchararray import center
x,y = make_blobs(
    n_samples = 300, centers = 4, cluster_std = 0.60, random_state = 42
)
```

```python
kmeans = KMeans(n_clusters=4,init="k-means++",max_iter=300,random_state=42)
```

```python
kmeans.fit(x)
```

```python
y = kmeans.predict(x)
```

```python
centroids = kmeans.cluster_centers_
```

```python
inertia = kmeans.inertia_
```

```python
print(inertia)
```

```python
plt.scatter(x[:, 0], x[:, 1], c=y, s=50, cmap="viridis", alpha=0.7, label="Data Points")
plt.scatter(
    centroids[:, 0],
    centroids[:, 1],
    c="red",
    s=200,
    marker="X",
    edgecolor="black",
    label="Centroids",
)
plt.title("K-Means Clustering via Scikit-Learn")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```
